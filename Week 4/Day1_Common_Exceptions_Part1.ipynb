## Exceptions
An exception is an error that occurs at runtime and interrupts the normal flow of your program.

### Types of exceptions in python

### 1. ZeroDivisionError

```python
a = 5
b = 0
print(a / b)
```

```python
try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    print(a / b)
except Exception:
    print("Cannot divide by zero!")
```

### 2. ValueError
Invalid value for a function/operation

```python
a = int(input("Enter numerator: "))
```

```python
# ValueError
age = int("ten")
```

```python
try:
    age = int("10")
    print("Age is:", age)
except ValueError:
    print("Cannot convert 'ten' to an integer. Please enter a number.")
```

### 3. TypeError
Incompatible types in an operation

```python
# TypeError
result = "Score: " + 90
```

```python
try:
    result = "Score: " + 90
except TypeError:
    print("TypeError: Cannot concatenate string with integer.")
```

### 4. NameError
Variable is not defined

```python
print(x)
```

```python
try:
    print(z)
except NameError:
    print(" NameError: Variable 'x' is not defined.")
```

### 5. IndexError
Invalid index in a list/tuple

```python
lst = [1, 2, 3]
lst[5]
```

```python
try:
    lst = [1, 2, 3]
    print(lst[5])
except IndexError:
    print("List index is out of range. Plese enter index withing range")

print("Hello")
print(5+5)
```
