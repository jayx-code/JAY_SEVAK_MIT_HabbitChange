### Using Lambda with map()

The map() function applies a given function to all items in an iterable (like a list) and returns a map object (which can be converted to a list):

```python
numbers = [1, 2, 3,3, 4]
squared = map(lambda x: x ** 2, numbers)
print(list(squared))
```

### Using Lambda with filter()


The filter() function constructs an iterator from elements of an iterable for which a function returns true:

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = filter(lambda x: x % 2 == 0, numbers)
```

```python
print(even_numbers)
```

```python
type(even_numbers)
```

```python
print(list(even_numbers))
```

### Using Lambda for String Manipulation

```python
# Convert a string to uppercase
uppercase = lambda s: s.upper()
print(uppercase('hello'))
```

```python
# Reverse a string
reverse = lambda s: s[::-1]
print(reverse('hello'))
```

### Enumerate function

The enumerate() function in Python is a built-in function that allows you to iterate over a list (or other iterable) and have an automatic counter. This function adds a counter to an iterable and returns it as an enumerate object. This can be very useful when you need to access both the index and the value of each item in an iterable.

```python
students = [("Virat", 85), ("Sachin", 92), ("Bumrah", 78), ("Dravid", 90)]
```

```python
# Use enumerate to iterate over the list with an index
for index, (name, score) in enumerate(students,start=5):
    print("Student",index,":",name,"scored", score)
```

```python
for index, (name, score) in enumerate(students, start=1):
    print(f"Student {index}: {name} scored {score}")
```

```python
for index, (name, score) in enumerate(students, start=1):
    print("Student {}: {} scored {}".format(index, name, score))
```

### Another example of enumerate

```python
students1 = ["A","B","C"]
for index,i in enumerate(students1,start=1):
    print("Student",index,":",i)
```

### Zip function

The zip() function in Python is a built-in function that allows you to combine two or more iterables (like lists or tuples) element-wise. It creates an iterator that aggregates elements from each of the iterables. The iterator stops when the shortest input iterable is exhausted.

```python
names = ["Virat", "Sachin", "Bumrah","Dravid"]
scores = (85, 92, 78)

# Use zip to combine the lists
combined = zip(names, scores)
```

```python
# Convert the zip object to a list and print it
combined_list = list(combined)
print(combined_list)
```

```python
combined_list[0]
```

### Star pattern questions

```python
n=10
for i in range(0,n):
    for j in range(0, i+1):
        print("*",end=" ")
    print("\r")
```

```python
n=int(input("Enter nuber of rows:"))
for i in range(n,0,-1):
    for j in range(n-i):
        print(' ',end='')
    for j in range(2*i -1):
        print('*', end='')
    print()
```

```python
n=9
i=0
while i<n:
    print(' '*(n-i-1)+'A '*(i+1))
    i+=1
```

### ATM machine code

```python
bal = 1000
k = 1

def bank_balance(trn_type, amount, bal):
    if trn_type == 1:
        print("Your Balance:", bal)
    elif trn_type == 2:
        bal += amount 
        print("Your Balance after Deposit:", bal)
    elif trn_type == 3:
        bal -= amount
        print("Your Balance after Withdrawal:", bal)
    return bal

while k <= 3:
    trn_type = int(input("Enter Selection 1. Balance, 2. Deposit, 3. Withdraw, 4. Exit: "))
    
    if trn_type in (1, 2, 3, 4):
        if trn_type == 4:
            print("Exiting...")
            break
        
        amt = 0
        if trn_type in (2, 3):
            amt = int(input("Enter Amount: "))
        
        bal = bank_balance(trn_type, amt, bal)
        print("Remaining Balance:", bal)
        k += 1
    else:
        print("Invalid Selection")
```
