# if, if else, if  elif else 

What is if-else?

In Python, if-else is used to make decisions. It allows you to execute different blocks of code based on conditions.

### Basic Syntax

```python
if condition:
    # block of code if condition is True    
elif another_condition:
    # block of code if another_condition is True    
else:
    # block of code if none of the above is True    
```

### Scenario 1: Simple if

```python
x=95
if (x>0):    
    print('x is positive')
```

```python
age = 18

if age >= 18:
    print("You are eligible to vote.")
```

### Scenario 2: if-else

```python
x=-95
if (x>0):
    print('x is positive')
else:
    print('x is negative')
```

```python
marks = 45

if marks >= 50:
    print("Pass")
else:
    print("Fail")
```

```python
#If else 
x=40
y=40
if x==y: #Boolean Expression
    print('x is equal to y') #Statement
else:
    print('x is not equal to y') #Statement
    print("ABC")
```

### Scenario 3: if-elif-else

```python
score = 85

if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 60:
    print("Grade C")
else:
    print("Fail")
```

### Scenario 4: Nested if

```python
age = 25
citizen = "India"

if age >= 18:
    if citizen == "India":
        print("You can vote.")
    else:
        print("Only Indian citizens can vote.")
```

```python
#No output
x=-5
if x<0:
    print('x is negative')
    print("Python class")
else:
    print("x")
    print("Hello")
print(1+2)
```

```python
#Multiple elif statements
price = 200

if price > 100:
    print("price is greater than 100")
elif price == 200:
    print("price is 100")
elif price < 100:
    print("price is less than 100")
```

### Scenario 5: Multiple if 

```python
#Multiple if statements
price = 200

if price >= 200:
    print("price is greater than 100")
if price == 200:
    print("price is 100")
if price < 200:
    print("price is less than 100")
```

```python
a = int(input("Enter a number: "))
if a<35:
    print("Less than 35")
if a<50:
    print("Less than 50")
elif a<75:
    print("less than 75")
else:
    print("Greater than 75")
    print("Number is too big")
print("This was fine")
```

```python
salary=int(input("Enter Salary"))
if salary < 1000:
    print("Salary is good")
    if salary < 500:
        print("I will buy nano car")
        if salary < 400:
            print("salary is medium")
        else:
            print("Salary is not so good")
    elif salary < 999:
        print("I will buy Maruti")
    else:
        print("I will buy BMW")
elif salary < 50000:
    print("salary is fine")
else:
    print("salary is low")
print("just for fun")
```

| Concept             | Example                    | Notes                                |
| ------------------- | -------------------------- | ------------------------------------ |
| Conditions          | `if x > 10:`               | Use `==`, `!=`, `<`, `>`, `>=`, `<=` |
| Multiple conditions | `if x > 10 and y < 5:`     | Use `and`, `or`, `not` for combining |
| Indentation         | Required in all blocks     | 4 spaces is standard                 |
| Use of colon `:`    | After `if`, `elif`, `else` | Don't forget it!                     |


### What is input()?

The input() function is used to take input from the user at runtime.

By default, input() returns a string, even if the user enters a number.

### Scenario 1: Taking a name

```python
name = input("Enter your name: ")
print("Welcome", name)
```

### Scenario 2: Taking a number (type conversion required)

```python
age = input("Enter your age: ")
print("You are", age, "years old")
type(age)
```

Type Conversion with int(), float()

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print("Sum:", num1 + num2)
```

```python
price = float(input("Enter price of item: "))
tax = float(input("Enter tax amount: "))
print("Total:", price + tax)
```
