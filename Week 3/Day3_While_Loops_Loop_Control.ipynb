# While loop

What is a while loop?

A while loop is used to repeat a block of code as long as a condition is True.

Use while when you don’t know in advance how many times the loop should run.

```python
while [boolean expression]:
    statement1
    statement2
    ...
    statementN
```

```python
# Will continue as long as the condition is true
i=1
while i<5:
    print(i)
    i=i+1
print('while loop has ended')    
```

```python
#Infinite Loop
i=1
while True:
    print(i,end=' ')
    i=i+1
```

### Using Break statement

```python
# Break (Breaks Out of loop)
i=1
while True:
    if i==10:
        break # It continues as long as if condition is False it stops when the if condition is True
    print(i)
    i=i+1
print("Hello")
```

### using Continue statement

```python
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

```python
#Continue (Takes back to loop)
i=1
while i<=10:
    if i==6:
        i=i+1 
        print("Hi",end=' ')        
        continue
    else:
        print(i,end=' ')
  
    i=i+1
print("Exit",end=' ')
```

### Real time example

```python
password = ""
while password != "admin123":
    password = input("Enter password: ")
print("Access granted")
```

| `for` loop                               | `while` loop                           |
| ---------------------------------------- | -------------------------------------- |
| Used when number of iterations is known  | Used when condition depends on runtime |
| Works well with sequences (lists, range) | Works well for conditional repetition  |
| Easier to read for fixed loops           | Flexible for unknown repetitions       |


### Naming a variable

Can start only with underscore or alphabet. Cannot contain any special character except underscore

| Rule # | Rule Description                                              |  Valid Examples       |  Invalid Examples      |
| ------ | ------------------------------------------------------------- | --------------------- | ---------------------- |
| 1      | Names must start with a **letter** or **underscore `_`**      | `name`, `_value`      | `1name`, `@score`      |
| 2      | Can only contain **letters**, **digits**, and **underscores** | `age_25`, `marks_100` | `first-name`, `total%` |
| 3      | **Cannot start with a number**                                | `data1`, `val_2`      | `1data`, `2value`      |
| 4      | **Case-sensitive** (uppercase ≠ lowercase)                    | `Name` ≠ `name`       | —                      |
| 5      | **No spaces allowed** in variable names                       | `student_name`        | `student name`         |
