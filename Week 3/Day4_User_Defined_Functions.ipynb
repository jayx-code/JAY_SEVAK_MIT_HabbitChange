# Functions

Functions are useful for breaking up a large program to make it easier to read and maintain.

They are also useful to avoid writing the same code at several different points in your program.

Function is a block of code that performs a specific task. 

This helps in avoiding redundancy in code

### Without function

```python
length1 = 5
breadth1 = 3
area1 = length1 * breadth1
print("Area 1:", area1)
```

```python
length2 = 10
breadth2 = 4
area2 = length2 * breadth2
print("Area 2:", area2)
```

### With Function

```python
def area_of_rectangle(length, breadth):
    return length * breadth
```

```python
print("Area 1:", area_of_rectangle(5, 3))
```

```python
print("Area 2:", area_of_rectangle(10, 4))
```

```python
# Function to determine if a no is in given range or not --> One Parameter
def as_range(x):
    if (x>50 and x<=100):
        return('It is in range')
    else:
        return('not in range')
```

```python
# Function to find max of 3 nos --> 3 parameters
def max3(a,b,c):
    if (a>b and a>c):
        return a
    elif (b>a and b>c):
        return b
    else: 
        return c
```

```python
max3(5,1,2)
```

### Write a function that accept a number and return the table of the same

```python
def table_of_n(n):
    for i in range(1,15):
        print(n," * ",i," = ",i*n)
```

```python
table_of_n(6)
```

### Write a function that counts number of vowel and consunants in a string/word.

```python
def vovcon(st):
    vov = 0
    con = 0
    for i in st:
        if i in ("AIEOUaieou"):
            vov += 1
        else:
            con += 1
    print("Vowels:", vov, " and consonants: ", con)
     #return vov,con
```

```python
a=vovcon("Hello")
a
```

### Default Arguments

```python
#Keyword Parameters --> Argument is keyword
def printdt(product,cost):
    print('Product:',product)
    print('Cost:',cost)
```

```python
printdt(product='keyboard',cost=2500)
```

```python
printdt(cost=44, product='Mouse')
```

```python
printdt('Chocolate', 20)
```

```python
printdt(20, "Chocolate")
```

```python
printdt("Chocolate")
```

```python
#Default Parameter. Default value is assigned in the definition

def sal_HRA(HRA,x=5):
    return(x*(HRA+1))
```
