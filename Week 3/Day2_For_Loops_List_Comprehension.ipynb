# For Loop

What is a for loop?
A for loop is used to repeat a block of code for each item in a sequence (like a list, string, tuple, etc.).

It is ideal when:

You know how many times you want to loop

You're looping through a collection

### Loop through a String

```python
#Traversing the string
s="republic"
for letter in s:
    print(letter)
```

```python
#Traversing the string
s="republic"
for letter in s:
    print(letter, end = ' ')
```

### Loop through a list

```python
#Traversing the list
L=['Naina','Pramila','Ashok','Rahul']
for name in L:
    print('Happy Friendship Day ' , name )
```

### Loop through a tuple

```python
#Traversing the Tuple
a=("apple","banana","republic")
for element in a:
    print(element, end='_')
```

### Loop through a dictionary

```python
#Traversing the Dictionary -->Key value pair
dict={1:"apple",2:"banana",3:"republic"}
for i,j in dict.items():
    print(i,' ', j)
```

```python
#Traversing the Dictionary -> Print only keys
dict={1:"apple",2:"banana",3:"republic"}
for a in dict.keys():
    print(a)
```

```python
# Traversing the Dictionary --> print only Values
dict={1:"apple",2:"banana",3:"republic"}
for fruit in dict.values():
    print(fruit)
```

### Using range() for numbers

```python
for i in range(5):
    print(i)
```

```python
#For loop with range to print nos 1 to 5
for i in range(1,5):
    print(i)
```

```python
#For loop with range to print nos 1 to 5
for i in range(1,5,2):
    print(i)
```

| Syntax              | Description         | Example                           |
| ------------------- | ------------------- | --------------------------------- |
| `range(n)`          | 0 to n-1            | `range(5)` → 0–4                  |
| `range(a, b)`       | a to b-1            | `range(2, 6)` → 2–5               |
| `range(a, b, step)` | a to b-1 with steps | `range(1, 10, 2)` → 1, 3, 5, 7, 9 |


### Print numbers from 10 to 1 (in reverse)

```python
for i in range(10, 0, -1):
    print(i)
```

```python
#Print the numbers from 100 to 200 which are divisible by 2
for i in range(100,201):
    if i %2 == 0:
        print(i, end=' ')
```

```python
for i in range(100,201,2):
    print(i, end=' ')
```

### Create a program to get the square of each number in a list

```python
numbers = [0,3,5,6,7,1]
res = []
for i in numbers:
    res.append(i * i)
print(res)
```

```python
numbers = []
for i in range(1,8):    
    numbers.append(i**2)
print(numbers)
```

### For loop through List Comprehension
List comprehension is a concise way to create a new list by applying an expression to each item in an existing iterable (like a list, range, or string).

It's a shortcut for writing for loops that create lists.



#### Example 1

Create list of squares from 1 to 5

```python
squares = [x ** 2 for x in range(1, 6)]
print(squares)
```

#### Example 2: Create a list of even numbers from 1 to 10

```python
evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)
```
