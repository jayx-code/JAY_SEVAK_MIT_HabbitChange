## Python Type Conversion

Type Conversion refers to converting one data type into another.

| Function    | Description                |
| ----------- | -------------------------- |
| `int()`     | Converts to integer        |
| `float()`   | Converts to float          |
| `str()`     | Converts to string         |
| `bool()`    | Converts to boolean        |
| `complex()` | Converts to complex number |


### String to Integer

```python
num_str = "100"
num = int(num_str)
print(num + 50)      
print(type(num))     
```

### Integer to String

```python
age = 25
str(age)
```

```python
type(age)
```

### Float to Integer (loses decimal part)

```python
score = 89.67
print(int(score))
```

### Integer to Float

```python
x = 5
print(float(x)) 
```

### Boolean to Integer

```python
print(int(True))  
print(int(False))  
```
