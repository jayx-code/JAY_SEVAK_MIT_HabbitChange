## Python Variables and Data Types

In Python, variables are used to store data. Python is dynamically typed, which means you don't need to declare the data type—Python automatically detects it!

## 1. Integer (int)

Whole numbers, positive or negative, without decimals.

```python
age = 25
print(age)
```

Python provides a built-in function type() to check the type of a variable.

```python
type(age)
```

## 2. Float (float)
Numbers with a decimal point.

```python
temperature = 36.6
print(temperature)
```

```python
type(temperature)
```

## 3. Boolean (bool)
Only two values: True or False.

```python
is_passed = True
print(is_passed)
```

```python
type(is_passed)
```

```python
print(10 > 5)  
```

```python
type(10 > 5)
```

## 4. String (str)
Textual data enclosed in single or double quotes. 

```python
name = "Virat Kohli"
print(name)
```

```python
type(name)
```

## 5. Complex Number (complex)
Numbers with a real and imaginary part.

```python
z = 3 + 4j
print(z)
```

```python
type(z)
```

## Exercise

```python
a = 10
b = 3.14
c = False
d = "Python"
e = 2 + 5j

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

## Print function

| Feature        | Example                       | Output                        |
| -------------- | ----------------------------- | ----------------------------- |
| Simple print   | `print("Hello")`              | Hello                         |
| Print variable | `print(name)`                 | Shubham (if name = "Shubham") |
| Multiple items | `print("Age:", age)`          | Age: 25                       |
| Separator      | `print("A", "B", sep="-")`    | A-B                           |
| End parameter  | `print("Loading", end="...")` | Loading...                    |


```python
print("Hello, Data Science!")
```

```python
name = "Virat"
age = 30
print(name)
print(age)
```

### Using + for Concatenation (Strings Only)

```python
first = "Data"
second = "Science"
print(first + " " + second)
```

### Customizing print() with sep and end

```python
print("Python", "Java", "R")
```

```python
print("Python", "Java", "R", sep=" | ")
```

```python
print("Loading", end="...")
```

### Print with formatted string

```python
name = "Vivek"
age = 25

print(f"My name is {name} and I am {age} years old.")
```

```python
a = 10
b = 5

print(f"Sum of {a} and {b} is {a + b}")
```
