### 3. Assignment Operators
Used to assign values to variables.

| Operator | Description         | Example                |
| -------- | ------------------- | ---------------------- |
| `=`      | Assign              | `a = 10`               |
| `+=`     | Add and assign      | `a += 5` => `a = a + 5`|
| `-=`     | Subtract and assign | `a -= 2`               |
| `*=`     | Multiply and assign | `a *= 3`               |
| `/=`     | Divide and assign   | `a /= 2`               |
| `//=`    | Floor divide assign | `a //= 3`              |
| `%=`     | Modulus and assign  | `a %= 3`               |
| `**=`    | Exponent and assign | `a **= 2`              |


```python
a = 10
a += 5
print(a)
```

```python
a *= 2
print(a)
```

### 4. Logical Operators
Used to combine conditional statements.

| Operator | Description                  | Example            | Result                |
| -------- | ---------------------------- | ------------------ | --------------------- |
| `and`    | True if both are True        | `x > 5 and x < 10` | True if both are true |
| `or`     | True if at least one is True | `x > 5 or x < 3`   | True                  |
| `not`    | Reverses the result          | `not(x > 5)`       | False if `x > 5`      |


```python
x = 7
print(x > 5 and x < 10)
```

```python
print(x < 5 or x == 7)
```

```python
print(not(x == 7))
```
