## Dealing with dates
Using Python’s built-in datetime module

```python
import datetime as dt
```

#### Get Today’s Date

```python
today = dt.date.today()
print("Today's Date:", today)
```

#### Get Current Date and Time

```python
now = dt.datetime.now()
print("Now:", now)
```

#### Extract Year, Month, Day

```python
print("Year:", now.year)
print("Month:", now.month)
print("Day:", now.day)
```

#### Format Dates (using strftime())

```python
formatted = now.strftime("%d-%m-%Y %H:%M:%S")
print("Formatted:", formatted)
```

#### Create a Specific Date

```python
dob = dt.date(2000, 5, 15)
print("DOB:", dob)
```

```python
date1=dt.datetime(2024, 8, 18, 10, 30)
print(date1)
```

### Differnce between 2 dates

```python
future = dt.date(2026, 12, 31)
days_left = future - today
print("Days until 2026 ends:", days_left.days)
```

```python
dob = dt.date(2001, 9, 20)
today = dt.date.today()

age_days = today - dob
print("You are", age_days.days, "days old.")
```

### Comments
Comments are lines in code that Python ignores during execution.

#### 1. Single line comments

```python
# This is a single-line comment
name = "Virat Kohli"  # This is also a comment
```

```python
# Calculate square of a number
num = 4
square = num ** 2  # num raised to power 2
print(square)
```
