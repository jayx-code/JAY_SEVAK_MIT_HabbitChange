### 5. Membership Operators
Used to test if a value is in a sequence (like list, string, tuple).

| Operator | Description                  | Example             | Result |
| -------- | ---------------------------- | ------------------- | ------ |
| `in`     | True if value exists         | `"a" in "data"`     | True   |
| `not in` | True if value does not exist | `"z" not in "data"` | True   |


```python
text = "python"
print("p" in text)   
```

```python
print("x" not in text)  
```

```python
x = 'IIT Academy'  
print('Cad' in x)
```

```python
print('cad' in x)
```

### 6. Identity Operators
Used to compare memory locations of two objects.

| Operator | Description                       | Example      | Result |
| -------- | --------------------------------- | ------------ | ------ |
| `is`     | True if both refer to same object | `x is y`     | True   |
| `is not` | True if not same object           | `x is not y` | True   |


```python
a = [5,2]
b = a
c = [5,2]

print(a is b)  #True because same memory
```

```python
print(a is c)     # False (same value, different memory)
```

```python
print(a == c)     # True (values are same)
```

```python
a2 = 'IITACADEMY'
b2 = 'iitacademy'
```

```python
print(a2 is b2)
```
