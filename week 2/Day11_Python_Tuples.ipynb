## Tuples

A tuple is:

An ordered, immutable collection of values

Defined using parentheses ( )

Can store multiple data types (int, float, string, etc.)

Cannot be changed (unlike lists)

### How to Create a Tuple

```python
# Simple tuple
t1 = (10, 20, 30)
t1
```

```python
# Mixed data types
t2 = ("Virat", 21, True)
t2
```

```python
# Single item tuple -> add a comma!
t4 = (10,)   # Not just (10)
t4
```

```python
# Empty tuple
t5 = ()
t5
```

### By passing a list to the 'tuple' function

```python
t1=tuple([1,2,3,2,3,"a",True])
t1
```

Accessing items of tuple

Indexing a tuple (Similar to string and list)

```python
t=(5,2,4,1,8)
t[2]
```

```python
t[-1]
```

Slicing a Tuple (Similar to string and list)

```python
t=('blue','red','green','yellow')
t[0:2]
```

Tuple is not mutable (immutable)

```python
t=('blue','red','green','yellow')
t[0]='white'
```

Finding Unique elements in a tuple using set function. Here output is a set data type. This can be converted to tuple by passing the set to tuple function

```python
t=(1,1,2,2)
t=set(t)
t
```

```python
# Converting set to tuple
t=tuple(t)
t
```

### Tuple Packing and Unpacking

```python
# Packing
student = ("Shubham", 21, "Data Science")
student
```

```python
# Unpacking
name, age, course = student
print(name)   
print(age)   
print(course) 
```

| Operation       | Code Example    | Result                    |
| --------------- | --------------- | ------------------------- |
| Create          | `t = (1, 2, 3)` | `(1, 2, 3)`               |
| Access          | `t[0]`, `t[-1]` | First/Last item           |
| Slice           | `t[1:3]`        | Subset of tuple           |
| Count value     | `t.count(2)`    | Number of times 2 appears |
| Find index      | `t.index(3)`    | Index of value 3          |
| Check existence | `2 in t`        | `True` or `False`         |
| Tuple length    | `len(t)`        | Total number of items     |
