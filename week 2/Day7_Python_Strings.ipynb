## 1. Strings
A string is a sequence of characters enclosed in quotes.
In Python, strings are immutable (cannot be changed once created).

### Ways to Create Strings

```python
# Using single quotes
s1 = 'Hello'
s1
```

```python
# Using double quotes
s2 = "World"
s2
```

```python
s3 = str("Hello bcd")
s3
```

```python
print("90"*3)
```

### String Indexing
Each character in a string has an index starting from 0.

```python
text = "Python for analytics"
print(text[0]) 
```

```python
print(text[8]) 
```

```python
#Negative indexing
print(text[-1])
print(text[-2])
```

### Common String Functions

```python
#Len function to find the length of string
text = "Python for analytics"
len(text)
```

```python
#Lower function to convert into lowercase
print(text.lower())  
```

```python
#upper function to convert into uppercase
print(text.upper()) 
```

```python
#Title function to convert first letter of each word in uppercase
s1 = "Virat is a good batsman"
s1.title()
```

```python
#Strip function will remove the leading and trailing spaces
s = "  hello  "
print(s.strip())
```

```python
#Lstrip will remove the leading spaces only.
print(s.lstrip())
```

```python
#Rstrip will remove the trailing spaces only.
print(s.rstrip())  
```

```python
#Replace function will replace the specific characters in the string
text = "Data Science"
print(text.replace(" ", "-"))   # Data-Science
```

```python
s="Python for data Science Python"
s=s.replace("P","S")
s
```

```python
#find will return the index position of the element
text = "machine"
print(text.find("c"))    # 2
print(text.find("z"))    # -1 (not found)
```

```python
s='python for data science'
t=s.find('n')# Find first occurence, returns index
t
```

```python
t1=s.find('n', 6)
t1
```

```python
#Index function
print(text.index("c"))   # 2
```

```python
print(text.index("z"))  
```

```python
#startswith is used to check if the string is starting with specific substring
text = "machine"
print(text.startswith("mac"))
```

```python
print(text.endswith("ine")) 
```

| Function       | Description                              | Example                      | Output            |
| -------------- | ---------------------------------------- | ---------------------------- | ----------------- |
| `len()`        | Length of string                         | `len("Hello")`               | `5`               |
| `lower()`      | Convert to lowercase                     | `"Python".lower()`           | `"python"`        |
| `upper()`      | Convert to uppercase                     | `"python".upper()`           | `"PYTHON"`        |
| `capitalize()` | Capitalize first letter                  | `"hello world".capitalize()` | `"Hello world"`   |
| `title()`      | Capitalize first letter of every word    | `"hello world".title()`      | `"Hello World"`   |
| `strip()`      | Remove leading and trailing spaces       | `"  abc  ".strip()`          | `"abc"`           |
| `lstrip()`     | Remove spaces from left side only        | `"  abc".lstrip()`           | `"abc"`           |
| `rstrip()`     | Remove spaces from right side only       | `"abc   ".rstrip()`          | `"abc"`           |
| `replace()`    | Replace part of string                   | `"a b".replace(" ", "-")`    | `"a-b"`           |
| `find()`       | Find position of first occurrence        | `"abcabc".find("b")`         | `1`               |
| `count()`      | Count number of occurrences of substring | `"banana".count("a")`        | `3`               |
| `split()`      | Split string into list                   | `"a b c".split()`            | `["a", "b", "c"]` |
| `join()`       | Join list into string                    | `".".join(["a","b"])`        | `"a.b"`           |
| `startswith()` | Checks if string starts with given value | `"data".startswith("da")`    | `True`            |
| `endswith()`   | Checks if string ends with given value   | `"data".endswith("a")`       | `True`            |
| `"a" in text`  | Membership test                          | `"Py" in "Python"`           | `True`            |
