Lists are mutable — you can change, add, or remove items after creation.

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [10, "hello", True, 3.14]
empty = []
```

#### Creating a list

```python
# 1. By Putting the items in a square bracket

lst=['apple', 'orange',1]
lst
```

```python
#2. Creating a blank list
lst=[]
lst
```

3. Using the list() Constructor --> It is also possible to use the list() constructor when creating a new list. We can pass following data types to the function 'list'. a)Tuple b)String c) Dictionary as a parameter to 'list' function

```python
lst=list()
lst
```

a) Passing 'tuple' to list() constructor

```python
#Note here we need to apply double round brackets
lst=list((1,2,3,4))
lst
```

b) Passing 'string' to list() constructor gives the output as list of characters

```python
a="Word"
l=list(a)
l
```

c) Passing a 'dictionary' to list() constructor gives the output of 'keys' of a dictionary

```python
a={'a':100,'b':200}
l=list(a)
l
```

```python
lst[2]
```

```python
lst[7]
```

Slicing a list

```python
#Lets consider the list as follows
lst = ["abc", 34, True, 40, "male"]
lst
```

```python
lst[1:3]
```

```python
lst[::-1]
```

```python
lst[0:4] #Here End index is not included
```

```python
# Indexing a Nested List (List within list)
lst=[1,2,'a',4,5,[40,10]]
lst
```

```python
lst[5][0]
```

```python
l=["Sachin", 1234, True, 45+8j, [5,3,1, "a","Virat",3.4,True], 6.7]
```

```python
l[0]
```

```python
l[0:3]
```

```python
l[4][2]
```

```python
l=["Arihant", 1234, True, 45+8j, [5,3,1, "a"], 6.7]
```

```python
for i in l:
    print(type(i))
```

```python
for i in l:
    if type(i)==list:
        for j in i:
            print(type(j))
```

```python
# Slicing in nested List
lst=[1,[8,9,10]]
lst[1:2] 
```

```python
# Nested List (For Loop)
lst=[1,2,'a',4,5,[40,10]]
for i in lst:
    print(i)
```

```python
for i in lst:
    print(i, end = ' ')
```
