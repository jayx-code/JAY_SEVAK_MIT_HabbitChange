## Dictionary

A dictionary is an unordered collection of key-value pairs.

Keys must be unique and immutable

Values can be of any type

Defined using curly braces {}



## Creating a Dictionary – Different Ways

### Method 1: Using Curly Braces

```python
student = {"name": "Shubham", "age": 21, "marks": 85}
student
```

### Method 2: Using dict() constructor and passing list of tuples

```python
student = dict([('A',100),('B',300)])
student
```

 ### Method 3: Empty Dictionary and Adding Items

```python
student = {}
student["name"] = "Shubham"
student["age"] = 21
student["marks"] = 85
student
```

#### For Loop using dictionary

```python
#For loop loops over the Keys
d={'A':100,'B':200,'C':300}
for i in d:
    print(i)
```

```python
d['B']
```

```python
#For looping over the values
d={'A':100,'B':200,'C':300}
for k in d:
    print(d[k])
```

```python
#For Looping over the items (key value pair)
for a,b in d.items():
    print(a,b)
```

| Operation        | Syntax                     | Example Result                  |
| ---------------- | -------------------------- | ------------------------------- |
| Create           | `d = {"a": 1, "b": 2}`     | `{'a': 1, 'b': 2}`              |
| Access           | `d["a"]`, `d.get("a")`     | `1`                             |
| Add / Update     | `d["c"] = 3`               | `{'a': 1, 'b': 2, 'c': 3}`      |
| Remove key       | `d.pop("a")`, `del d["b"]` | Removes that key                |
| All keys         | `d.keys()`                 | `dict_keys([...])`              |
| All values       | `d.values()`               | `dict_values([...])`            |
| All items        | `d.items()`                | `dict_items([...])`             |
| Clear all        | `d.clear()`                | `{}`                            |
| Merge other dict | `d.update(other_dict)`     | Adds all pairs from other\_dict |


### Nested Dictionary

```python
students = {
    "101": {
        "name": "Shubham",
        "age": 21,
        "marks": 85
    },
    "102": {
        "name": "Mayank",
        "age": 22,
        "marks": 90
    }
}

students
```

### Exercise

```python
product = {
    "id": 101,
    "name": "Laptop",
    "brand": "Dell",
    "price": 55000
}
product
```

### Tasks:
    
Update the price to 50000

Add a new key "in_stock" with value True

Remove the "brand" key

Print all key-value pairs

```python
# Step 1: Update the price
product["price"] = 50000
product
```

```python
# Step 2: Add a new key-value pair
product["in_stock"] = True
product
```

```python
# Step 3: Remove the brand key
product.pop("brand")
product
```

```python
# Step 4: Print all key-value pairs
print("Product Details:")
for key, value in product.items():
    print(key, ":", value)
```
