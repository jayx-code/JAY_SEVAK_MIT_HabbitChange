Difference of Lists --> Given 2 lists l1 & l2, Find the elements present in l1 but are absent in l2. This cannot be achieved by simple subtraction of two lists. It is a two step process 1) Convert both lists into set by using set function and subtract. It gives output in the set form. 2) Convert the set to list

```python
l1=[8,7,4,3,2,2] 
l2=[8,6,4,1,2]
list(set(l1)-set(l2))#Output in set 
```

```python
set(l1)
```

```python
set(l2)
```

```python
list(set(l1)-set(l2)) #To convert to list
```

| Function    | Description                    | Example             |
| ----------- | ------------------------------ | ------------------- |
| `len()`     | Number of items                | `len([1, 2])` → 2   |
| `append()`  | Add item to end                | `a.append(5)`       |
| `insert()`  | Add item at given index        | `a.insert(1, "hi")` |
| `remove()`  | Remove first matching value    | `a.remove(2)`       |
| `pop()`     | Remove by index (default last) | `a.pop()`           |
| `sort()`    | Sort in ascending order        | `a.sort()`          |
| `reverse()` | Reverse the list               | `a.reverse()`       |
| `count()`   | Count occurrences              | `a.count(2)`        |
| `index()`   | Get index of a value           | `a.index(3)`        |
| `extend()`  | Combine two lists              | `a.extend(b)`       |
| `clear()`   | Remove all items               | `a.clear()`         |


### Exercise on Lists

Shopping Cart Tracker

Q:
You are building a shopping cart system.
Start with a list of items: ["Milk", "Bread", "Butter"]

Perform the following:

Add "Eggs" to the cart

Remove "Milk" from the cart

Insert "Cheese" at the beginning

Show total items in the cart

```python
# Step 1: Start with initial cart
cart = ["Milk", "Bread", "Butter"]
print("Initial cart:", cart)
```

```python
# Step 2: Add "Eggs"
cart.append("Eggs")
print("After adding Eggs:", cart)
```

```python
# Step 3: Remove "Milk"
cart.remove("Milk")
print("After removing Milk:", cart)
```

```python
# Step 4: Insert "Cheese" at the beginning
cart.insert(0, "Cheese")
print("After inserting Cheese at beginning:", cart)
```

```python
# Step 5: Show total items in cart
print("Total items in cart:", len(cart))
```
