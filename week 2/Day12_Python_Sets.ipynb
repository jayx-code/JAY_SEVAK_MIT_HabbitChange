## 5. Sets
A set is:

An unordered, unindexed collection of unique elements

Defined using curly braces {} or the set() function

Automatically removes duplicates

Useful for membership testing and set operations like union, intersection, difference, etc.

### How to create a set

```python
# Using curly braces
fruits = {"apple", "banana", "cherry"}
fruits
```

```python
# Using set() function
#Passing a list in set function
numbers = set([1, 2, 3, 3, 4])
numbers
```

```python
#Passing a tuple in set function
a=set((1,2,3,4))
a
```

```python
# Empty set 
empty = set()       # Correct
not_a_set = {}      # This is an empty dictionary
```

```python
#No duplicates
s = {1, 2, 2, 3}
print(s)
```

```python
#Indexing and slicing do not work
s = {1, 2, 3}
s[0]   #will give an error
```

### Set operations

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))         # {1, 2, 3, 4, 5}
print(a.intersection(b))  # {3}
print(a.difference(b))    # {1, 2}
print(b.difference(a))    # {4, 5}
print(a.symmetric_difference(b))  # {1, 2, 4, 5}
```

| Operation       | Syntax                      | Example Result               |
| --------------- | --------------------------- | ---------------------------- |
| Create set      | `s = {1, 2, 3}`             | `{1, 2, 3}`                  |
| Add item        | `s.add(4)`                  | `{1, 2, 3, 4}`               |
| Add multiple    | `s.update([5, 6])`          | `{1, 2, 3, 4, 5, 6}`         |
| Remove item     | `s.remove(2)`               | Removes 2 or raises error    |
| Safe remove     | `s.discard(9)`              | No error if not found        |
| Combine sets    | `s.union(t)`                | All unique elements          |
| Common elements | `s.intersection(t)`         | Common items only            |
| Not common      | `s.symmetric_difference(t)` | Items in s or t but not both |


### Exercise

Use sets to manage and compare product categories in a store.

You run an online store that sells products in different categories. Two sets represent the categories available in two different warehouses:

```python
warehouse_A = {"laptops", "mobiles", "headphones", "chargers"}
warehouse_B = {"mobiles", "chargers", "tablets", "speakers"}
```

### Tasks:
Print all unique categories across both warehouses

Find the common categories

Find categories that are only in Warehouse A

Find categories that are in either A or B, but not both

Add "powerbanks" to Warehouse A

Remove "chargers" from Warehouse B

```python
warehouse_A = {"laptops", "mobiles", "headphones", "chargers"}
warehouse_B = {"mobiles", "chargers", "tablets", "speakers"}

# 1. All unique categories (Union)
print("All categories:", warehouse_A.union(warehouse_B))

# 2. Common categories (Intersection)
print("Common categories:", warehouse_A.intersection(warehouse_B))

# 3. Only in A (Difference)
print("Only in Warehouse A:", warehouse_A.difference(warehouse_B))

# 4. Either A or B but not both (Symmetric Difference)
print("Unique to each warehouse:", warehouse_A.symmetric_difference(warehouse_B))

# 5. Add a single item to Warehouse A
warehouse_A.add("powerbanks")
print("Warehouse A after adding powerbanks:", warehouse_A)

# 6. Remove 'chargers' from Warehouse B
warehouse_B.discard("chargers")
print("Warehouse B after removing chargers:", warehouse_B)
```
